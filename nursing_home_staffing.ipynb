```python
# Creating outlier mask, to remove, but make note, as it's just one row #
outlier_mask = df['Hrs_LPN_ctr'] == df['Hrs_LPN_ctr'].max()

removed_rows = df[outlier_mask]
removed_rows.to_csv('removed_outliers_log.csv', index=False)

df = df[~outlier_mask]
```

```python
# Shape after removing the one outlier row: (1325324, 33) -> (1325323, 33)
print(df.shape)
```

```python
# total nurse hours = RN + LPN + CNA
# (CMS's "total nurse staffing" adds nurse aide trainees (Hrs_NAtrn) and med aides (Hrs_MedAide).
#  Not included here; noted in documentation)
df['total_nurse_hours'] = df['Hrs_RN'] + df['Hrs_LPN'] + df['Hrs_CNA']
print(df[['Hrs_RN', 'Hrs_LPN', 'Hrs_CNA', 'total_nurse_hours']].head())
```

```python

#   - MDScensus == 0: no residents that day -> dividing by 0 gives inf
#   - 0 nurse hours with residents: a facility can't run like that, so these are likely missing reports
#   Note: these rows are dropped for ALL metrics below.
zero_census = df['MDScensus'] == 0
zero_hours = df['total_nurse_hours'] == 0
print('Days with 0 residents:', zero_census.sum())
print('Days with residents but 0 nurse hours:', (zero_hours & ~zero_census).sum())

df = df[~zero_census & ~zero_hours]
print(df.shape)
```

```python
# 1) Nurse Staffing per Resident
# HPRD = nurse Hours Per Resident Day (the CMS term). ~4 means each resident gets ~4 nursing hours a day.
df['nurse_hprd'] = df['total_nurse_hours'] / df['MDScensus']
print(df[['nurse_hprd']].head(10))
```

```python
df['MONTH'] = df['WorkDate'].dt.month
```

```python
# 2) Total Nurse Hours by facility / state / month
total_nurse_hours_facility = df.groupby(['PROVNUM', 'STATE', 'MONTH'])['total_nurse_hours'].sum().reset_index()
print(total_nurse_hours_facility.head(25))
```

```python
# 3) Contract vs Permanent Staff Ratio
df['ctr_staff_hours'] = df['Hrs_RN_ctr'] + df['Hrs_LPN_ctr'] + df['Hrs_CNA_ctr']
print(df[['Hrs_RN_ctr', 'Hrs_LPN_ctr', 'Hrs_CNA_ctr', 'ctr_staff_hours']].head())
```

```python
# no more NaN, since 0-hour rows were removed
df['contract_ratio'] = df['ctr_staff_hours'] / df['total_nurse_hours']
print(df[['contract_ratio']].head(25))
print('NaN contract ratios:', df['contract_ratio'].isna().sum())
```

```python
# Double checking contract vs permanent staff ratio
print(df[df['contract_ratio'] > 0].head())
```

```python
# 4) Average Patient Census (avg residents per facility OR per state)
avg_census_facility = df.groupby(['PROVNUM', 'STATE'])['MDScensus'].mean().reset_index()
avg_census_state = df.groupby('STATE')['MDScensus'].mean().reset_index()
print(avg_census_facility.head())
print(avg_census_state.head())
```

```python
# 5) Facilities with Highest Staffing Hours
facilities_with_highest_hours = df.groupby(['PROVNUM', 'PROVNAME', 'STATE'])['total_nurse_hours'].sum().reset_index()
facilities_with_highest_hours = facilities_with_highest_hours.sort_values('total_nurse_hours', ascending=False)
```

```python
# Facilities with highest staffing hours (top 10)
# total hours mostly reflects facility SIZE (biggest facilities rise to the top)
print(facilities_with_highest_hours.head(10))
```

```python
# What is the relationship between nurse staffing levels and patient census?

# corr runs -1 to +1. ~0.91 is a strong positive correlation:
# facilities with more residents report more nurse hours.
# This mostly reflects size (bigger facility = more hours), so it does NOT show a facility is well-run.
# Per-resident staffing (nurse_hprd) is the better measure of staffing quality.
df['total_nurse_hours'].corr(df['MDScensus'])
```

```python
# What are the average staffing levels by state? (raw hours per facility-day)
#  this ranking mostly reflects facility size (e.g., NY has large facilities)
avg_staffing_by_state = df.groupby('STATE')['total_nurse_hours'].mean().reset_index(name='average_hours')
avg_staffing_by_state = avg_staffing_by_state.sort_values('average_hours', ascending=False)
print(avg_staffing_by_state.head(10))
```

```python
#  Per-resident staffing by state (fair comparison regardless of facility size)
hprd_by_state = df.groupby('STATE')[['total_nurse_hours', 'MDScensus']].sum()
hprd_by_state['nurse_hprd'] = hprd_by_state['total_nurse_hours'] / hprd_by_state['MDScensus']
hprd_by_state = hprd_by_state.sort_values('nurse_hprd', ascending=False).reset_index()
print(hprd_by_state[['STATE', 'nurse_hprd']].head(10))
print(hprd_by_state[['STATE', 'nurse_hprd']].tail(10))
```

```python
# Which facilities have the lowest staffing coverage?
lowest_coverage = df.groupby(['PROVNUM', 'PROVNAME', 'STATE'])['nurse_hprd'].mean().reset_index()
lowest_coverage = lowest_coverage.sort_values('nurse_hprd', ascending=True)
print(lowest_coverage.head(10))
```

```python
# Which states rely most on contract staff?
#  weighted ratio = total contract hours / total nurse hours per state
# (averaging daily ratios gave a tiny facility's day the same weight as a huge one's)
state_ctr_reliance = df.groupby('STATE')[['ctr_staff_hours', 'total_nurse_hours']].sum()
state_ctr_reliance['contract_ratio'] = state_ctr_reliance['ctr_staff_hours'] / state_ctr_reliance['total_nurse_hours']
state_ctr_reliance = state_ctr_reliance.sort_values('contract_ratio', ascending=False).reset_index()
print(state_ctr_reliance[['STATE', 'contract_ratio']].head(10))
```
